# Investigating Evidence for Dark Matter Using Galactic Rotation Curves

## 1. Introduction

Galaxies contain stars, gas, and other forms of matter that contribute to their gravitational fields. A galactic rotation curve describes how the orbital velocity of matter varies with distance from the galactic centre. By comparing observed rotation velocities with theoretical predictions based on visible matter, it is possible to investigate whether the observed motion can be explained by the known baryonic components alone.

In many galaxies, the observed rotation curve differs from the prediction based on visible matter. This discrepancy is commonly interpreted as evidence for additional gravitating matter, with dark matter being one of the leading explanations.

In this project, rotation-curve data from the SPARC (Spitzer Photometry and Accurate Rotation Curves) database are analysed to compare observed velocities with velocities predicted from gas and stellar components. NGC 3198 is selected as the main case study because its rotation curve extends to large galactic radii and has been extensively studied. NGC 2683 and DDO 154 are included as comparison galaxies to investigate whether similar discrepancies occur in other galactic systems.

The main objective is to determine how well the observed rotation curves can be reproduced by the visible-matter contribution under the adopted model assumptions and to discuss the implications of any differences.

## 2. Objectives

1. To understand the relationship between galactic rotation curves and the distribution of mass within galaxies.
2. To calculate the predicted baryonic rotation curve using the gas, stellar-disk, and stellar-bulge contributions provided by SPARC.
3. To compare the predicted baryonic velocities with the observed rotation velocities of NGC 3198.
4. To analyse the velocity discrepancy and the ratio between observed and baryonic velocities.
5. To compare the results for NGC 3198, NGC 2683, and DDO 154.
6. To discuss the limitations of the analysis and the implications of discrepancies between observed and predicted velocities.

## 3. Theoretical Background

### 3.1 Galactic Rotation Curves

A galactic rotation curve represents the variation of orbital velocity with galactocentric radius. Observations of stars and gas at different radii provide information about the gravitational field of a galaxy and the distribution of its mass.

For a spherically symmetric mass distribution, the circular orbital velocity is related to the enclosed mass by

$$
V_c(r)=\sqrt{\frac{GM(<r)}{r}}
$$

where:

- $V_c(r)$ is the circular orbital velocity at radius $r$.
- $G$ is the gravitational constant.
- $M(<r)$ is the total mass enclosed within radius $r$.

This relation illustrates how orbital velocities can be used to investigate a galaxy's mass distribution. However, galaxies often contain flattened stellar and gaseous disks, so their complete gravitational fields cannot generally be described by the spherical approximation alone.

If the visible matter adequately explained a galaxy's gravitational field, the observed rotation curve would be consistent with the prediction from its gas and stellar components. A persistent difference between the observed and predicted velocities indicates a mass discrepancy that requires further interpretation.

### 3.2 Baryonic Contribution to the Rotation Curve

The visible matter considered in this analysis consists of gas, the stellar disk, and the stellar bulge. The SPARC database provides the corresponding velocity contributions, which can be combined to calculate the predicted baryonic rotation curve.

The baryonic velocity is calculated using

$$
V_{\mathrm{bar}}(r)=
\sqrt{
V_{\mathrm{gas}}(r)|V_{\mathrm{gas}}(r)|
+\Upsilon_{\mathrm{disk}}V_{\mathrm{disk}}^2(r)
+\Upsilon_{\mathrm{bul}}V_{\mathrm{bul}}^2(r)
}
$$

where:

- $V_{\mathrm{bar}}(r)$ is the predicted baryonic circular velocity.
- $V_{\mathrm{gas}}(r)$ is the gas contribution.
- $V_{\mathrm{disk}}(r)$ is the stellar-disk contribution.
- $V_{\mathrm{bul}}(r)$ is the stellar-bulge contribution.
- $\Upsilon_{\mathrm{disk}}$ is the adopted stellar-disk mass-to-light ratio.
- $\Upsilon_{\mathrm{bul}}$ is the adopted stellar-bulge mass-to-light ratio.

For this analysis, the following values are adopted:

$$
\Upsilon_{\mathrm{disk}}=0.5\;M_\odot/L_\odot
$$

$$
\Upsilon_{\mathrm{bul}}=0.7\;M_\odot/L_\odot
$$

The disk mass-to-light ratio of $0.5\;M_\odot/L_\odot$ is commonly adopted in SPARC mass modelling at 3.6 micrometres. The bulge value of $0.7\;M_\odot/L_\odot$ is the assumption used in this analysis. These values are model assumptions rather than universal constants.

The gas contribution is written as $V_{\mathrm{gas}}|V_{\mathrm{gas}}|$ to preserve its sign. The component contributions are combined inside the square root; the component velocities are not simply added together.

The resulting baryonic rotation curve represents the predicted contribution of the modelled visible matter under the adopted assumptions. It is then compared with the observed rotation curve to investigate any discrepancies.

### 3.3 Velocity Discrepancy

To investigate whether visible matter can reproduce the observed rotation curve, the calculated baryonic velocity is compared with the observed velocity at each measured radius.

The velocity discrepancy is defined as

$$
\Delta V(r)=V_{\mathrm{obs}}(r)-V_{\mathrm{bar}}(r)
$$

where $V_{\mathrm{obs}}(r)$ is the observed rotation velocity and $V_{\mathrm{bar}}(r)$ is the predicted baryonic velocity.

A discrepancy close to zero indicates that the baryonic prediction approximately reproduces the observed velocity. A positive discrepancy indicates that the observed velocity exceeds the prediction, whereas a negative discrepancy indicates that the baryonic prediction is higher than the observed value.

A positive discrepancy may be consistent with an additional gravitational contribution from matter not accounted for in the visible-matter model. However, the discrepancy alone does not prove the existence of dark matter, because the result also depends on observational uncertainties and the adopted model assumptions.

### 3.4 Observed-to-Baryonic Velocity Ratio

The relative difference between the observed and predicted velocities is also examined using the velocity ratio:

$$
R_V(r)=\frac{V_{\mathrm{obs}}(r)}{V_{\mathrm{bar}}(r)}
$$

A ratio close to one indicates agreement between the two velocities. A ratio greater than one indicates that the observed velocity exceeds the baryonic prediction, while a ratio below one indicates that the baryonic prediction exceeds the observed velocity.


## 4. Data and Methodology

### 4.1 Dataset Description

The rotation-curve data used in this project are obtained from the SPARC database, which provides observed rotation velocities and baryonic component contributions for disk galaxies. The dataset allows the observed rotational motion to be compared with the predicted contribution from visible matter.

The main analysis focuses on NGC 3198. NGC 2683 and DDO 154 are subsequently analysed using the same calculation procedure to provide comparisons between galaxies with different structures and radial coverage.

The principal data columns used in the analysis are:

| Column | Description | Unit |
|---|---|---|
| `Rad` | Galactocentric radius | kpc |
| `Vobs` | Observed rotation velocity | km/s |
| `errV` | Uncertainty in observed velocity | km/s |
| `Vgas` | Gas contribution to the rotation curve | km/s |
| `Vdisk` | Stellar-disk contribution | km/s |
| `Vbul` | Stellar-bulge contribution | km/s |
| `SBdisk` | Stellar-disk surface brightness | Dataset-dependent units |
| `SBbul` | Stellar-bulge surface brightness | Dataset-dependent units |

The surface-brightness columns provide additional information about the stellar components but are not directly required in the baryonic velocity calculation performed here.

### 4.2 Calculation Procedure

The analysis follows these steps:

1. Load the rotation-curve data for NGC 3198 and inspect the relevant columns.
2. Calculate the baryonic velocity at each measured radius using the adopted mass-to-light ratios.
3. Plot the observed rotation curve, including observational error bars, alongside the predicted baryonic rotation curve.
4. Calculate and plot the velocity discrepancy as a function of radius.
5. Calculate the observed-to-baryonic velocity ratio to examine the relative difference between the two curves.
6. Repeat the calculations for NGC 2683 and DDO 154.
7. Compare the results and discuss the limitations of the visible-matter model.

The same equation and mass-to-light ratio assumptions are used for all three galaxies to maintain a consistent comparison.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

file_path = Path("NGC3198_rotmod.dat")

if not file_path.exists():
    raise FileNotFoundError(
        "NGC3198_rotmod.dat was not found. "
        "Check the file path."
    )

columns = [
    "Rad", "Vobs", "errV", "Vgas",
    "Vdisk", "Vbul", "SBdisk", "SBbul"
]

df = pd.read_csv(
    file_path,
    sep=r"\s+",
    comment="#",
    names=columns
)

print("Rows:", len(df))
print("Columns:", df.columns.tolist())
print("Missing values:\n", df.isna().sum())
print(df.head())
print(df.tail())
```

```python

# Adopted stellar mass-to-light ratios
U_disk = 0.5
U_bul = 0.7

# Calculate the baryonic rotation velocity
df["Vbar_squared"] = (
    df["Vgas"] * np.abs(df["Vgas"])
    + U_disk * df["Vdisk"]**2
    + U_bul * df["Vbul"]**2
)

if (df["Vbar_squared"] < 0).any():
    raise ValueError("Negative squared baryonic velocity encountered.")

df["Vbar"] = np.sqrt(df["Vbar_squared"])

# Compare observed and baryonic velocities
df["Velocity_Discrepancy"] = df["Vobs"] - df["Vbar"]
df["Velocity_Ratio"] = df["Vobs"] / df["Vbar"]

display(
    df[[
        "Rad", "Vobs", "errV", "Vbar",
        "Velocity_Discrepancy", "Velocity_Ratio"
    ]].round(2)
)
```

## 5. Analysis of NGC 3198

### 5.1 Observed and Baryonic Rotation Curves

The first analysis compares the observed rotation curve of NGC 3198 with the rotation curve predicted from its baryonic components. The observed velocities and their measurement uncertainties are plotted against galactocentric radius. The predicted baryonic curve is overlaid to show how closely the visible-matter model reproduces the observed data.

The comparison is examined across the measured radial range, with particular attention to the outer regions of the galaxy. If the baryonic prediction falls below the observed rotation curve over an extended range, this indicates that the adopted visible-matter model does not fully reproduce the observed rotational motion.

The plot therefore provides a visual assessment of the agreement between the observed data and the predicted baryonic contribution. The interpretation must also account for the adopted mass-to-light ratios and the limitations of the model.

```python

plt.figure(figsize=(9, 5.5))

plt.errorbar(
    df["Rad"],
    df["Vobs"],
    yerr=df["errV"],
    fmt="o",
    capsize=3,
    label="Observed velocity"
)

plt.plot(
    df["Rad"],
    df["Vbar"],
    "s--",
    label="Baryonic prediction"
)

plt.xlabel("Radius (kpc)")
plt.ylabel("Rotation velocity (km/s)")
plt.title("NGC 3198: Observed vs Baryonic Rotation Curve")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

### 5.2 Velocity Discrepancy Analysis

The velocity discrepancy is calculated at each measured radius by subtracting the baryonic prediction from the observed velocity:

$$
\Delta V(r)=V_{\mathrm{obs}}(r)-V_{\mathrm{bar}}(r)
$$

The resulting values are plotted against galactocentric radius, with a horizontal reference line at zero. Values above zero indicate that the observed velocity exceeds the baryonic prediction, whereas values below zero indicate that the baryonic prediction is higher than the observed velocity.

The magnitude and radial behaviour of the discrepancy help identify the regions where the visible-matter model agrees with the observations and where it differs. A sustained positive discrepancy would be consistent with an additional gravitational contribution not accounted for by the adopted baryonic model.

The error bars represent the observational uncertainty in the measured rotation velocity. They do not include all uncertainties associated with the baryonic model, the adopted mass-to-light ratios, or other systematic effects.

```python

plt.figure(figsize=(9, 5))

plt.errorbar(
    df["Rad"],
    df["Velocity_Discrepancy"],
    yerr=df["errV"],
    fmt="o",
    capsize=3
)

plt.axhline(0, linestyle="--", color="black")

plt.xlabel("Radius (kpc)")
plt.ylabel("Observed − Baryonic velocity (km/s)")
plt.title("Velocity Discrepancy in NGC 3198")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

outer = df.iloc[-1]

print("Galaxy: NGC 3198")
print(f"Number of measurements: {len(df)}")
print(f"Outermost radius: {outer['Rad']:.2f} kpc")
print(f"Observed velocity: {outer['Vobs']:.2f} km/s")
print(f"Baryonic prediction: {outer['Vbar']:.2f} km/s")
print(f"Velocity discrepancy: {outer['Velocity_Discrepancy']:.2f} km/s")
print(f"Observed/baryonic velocity ratio: {outer['Velocity_Ratio']:.2f}")
```

## 6. Comparison with Other Galaxies

After analysing NGC 3198, the same procedure is applied to NGC 2683 and DDO 154. This comparison investigates whether discrepancies between observed and baryonic velocities are also present in galaxies with different structures and radial coverage.

The comparison includes the number of measurements, the radial range, the observed and predicted velocities, and the corresponding velocity discrepancies. The rotation curves are also plotted together with their individual baryonic predictions to allow their radial behaviour to be examined.

Since the galaxies are sampled at different radii and have different numbers of measurements, their outermost data points are not directly equivalent. A larger discrepancy at the outermost point of one galaxy does not necessarily mean that the galaxy contains more dark matter than another. The results must therefore be interpreted alongside each galaxy's full rotation curve and the assumptions used in the calculation.

This comparison helps assess whether the discrepancy observed in the main case study is also evident in the other systems analysed. It provides a limited comparison of three galaxies rather than a statistical conclusion about the entire SPARC sample.

```python

from pathlib import Path

file_path = Path("DDO154_rotmod.dat")

if not file_path.exists():
    raise FileNotFoundError(
        "DDO154_rotmod.dat not found. Check the file path."
    )

ddo154 = pd.read_csv(
    file_path,
    sep=r"\s+",
    comment="#",
    names=columns
)

ddo154["Vbar_squared"] = (
    ddo154["Vgas"] * np.abs(ddo154["Vgas"])
    + 0.5 * ddo154["Vdisk"]**2
    + 0.7 * ddo154["Vbul"]**2
)

if (ddo154["Vbar_squared"] < 0).any():
    raise ValueError(
        "Negative squared baryonic velocity found for DDO 154."
    )

ddo154["Vbar"] = np.sqrt(ddo154["Vbar_squared"])

ddo154["Velocity_Discrepancy"] = (
    ddo154["Vobs"] - ddo154["Vbar"]
)

ddo154["Velocity_Ratio"] = (
    ddo154["Vobs"] / ddo154["Vbar"].replace(0, np.nan)
)

outer = ddo154.iloc[-1]

print("Galaxy: DDO 154")
print("Measurements:", len(ddo154))
print(f"Outermost radius: {outer['Rad']:.2f} kpc")
print(f"Observed velocity: {outer['Vobs']:.2f} km/s")
print(f"Baryonic prediction: {outer['Vbar']:.2f} km/s")
print(f"Velocity discrepancy: {outer['Velocity_Discrepancy']:.2f} km/s")
print(f"Observed/baryonic ratio: {outer['Velocity_Ratio']:.2f}")
```

```python
ngc2683 = pd.read_csv(
    "NGC2683_rotmod.dat",
    sep=r"\s+",
    comment="#",
    names=columns
)

ngc2683["Vbar_squared"] = (
    ngc2683["Vgas"] * np.abs(ngc2683["Vgas"])
    + 0.5 * ngc2683["Vdisk"]**2
    + 0.7 * ngc2683["Vbul"]**2
)

ngc2683["Vbar"] = np.sqrt(ngc2683["Vbar_squared"])

ngc2683["Velocity_Discrepancy"] = (
    ngc2683["Vobs"] - ngc2683["Vbar"]
)

ngc2683["Velocity_Ratio"] = (
    ngc2683["Vobs"] / ngc2683["Vbar"]
)

print("Number of measurements:", len(ngc2683))
print("Outermost radius:", ngc2683["Rad"].iloc[-1], "kpc")
print("Observed velocity:", ngc2683["Vobs"].iloc[-1], "km/s")
print("Baryonic prediction:", round(ngc2683["Vbar"].iloc[-1], 2), "km/s")
print("Velocity discrepancy:", round(ngc2683["Velocity_Discrepancy"].iloc[-1], 2), "km/s")
print("Velocity ratio:", round(ngc2683["Velocity_Ratio"].iloc[-1], 2))
```

```python

galaxies = {
    "NGC 3198": df,
    "DDO 154": ddo154,
    "NGC 2683": ngc2683
}

comparison = []

for name, galaxy in galaxies.items():
    outer = galaxy.loc[galaxy["Rad"].idxmax()]

    comparison.append({
        "Galaxy": name,
        "Measurements": len(galaxy),
        "Outermost Radius (kpc)": outer["Rad"],
        "Observed Velocity (km/s)": outer["Vobs"],
        "Baryonic Prediction (km/s)": outer["Vbar"],
        "Velocity Discrepancy (km/s)": outer["Velocity_Discrepancy"],
        "Observed/Baryonic Ratio": outer["Velocity_Ratio"]
    })

comparison_df = pd.DataFrame(comparison).round(2)

print(comparison_df.to_string(index=False))
```

```python

galaxies = {
    "NGC 3198": df,
    "DDO 154": ddo154,
    "NGC 2683": ngc2683
}

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for ax, (name, galaxy) in zip(axes, galaxies.items()):
    ax.errorbar(
        galaxy["Rad"],
        galaxy["Vobs"],
        yerr=galaxy["errV"],
        fmt="o",
        capsize=3,
        label="Observed"
    )

    ax.plot(
        galaxy["Rad"],
        galaxy["Vbar"],
        "s--",
        label="Baryonic prediction"
    )

    ax.set_title(name)
    ax.set_xlabel("Radius (kpc)")
    ax.set_ylabel("Velocity (km/s)")
    ax.legend()
    ax.grid(alpha=0.3)

fig.suptitle(
    "Observed vs. Baryonic Rotation Curves",
    fontsize=14
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for name, galaxy in galaxies.items():
    plt.scatter(
        galaxy["Rad"],
        galaxy["Velocity_Discrepancy"],
        label=name,
        s=35
    )

plt.axhline(0, color="black", linestyle="--", linewidth=1)
plt.xlabel("Radius (kpc)")
plt.ylabel("Observed − Baryonic Velocity (km/s)")
plt.title("Velocity Discrepancy Across Three Galaxies")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```


### 7. Results and Discussion

The analysis of NGC 3198 shows that the observed rotation curve increasingly differs from the baryonic prediction across much of the measured outer region. At the outermost measured radius of 44.08 kpc, the observed velocity is 149.00 km/s, while the predicted baryonic velocity is 64.71 km/s. This gives a velocity discrepancy of 84.29 km/s and an observed-to-baryonic velocity ratio of approximately 2.30.

These results indicate that the adopted visible-matter model does not fully reproduce the observed rotation curve of NGC 3198 at large radii. The difference is consistent with an additional gravitational contribution not accounted for by the modelled gas and stellar components.

The comparison galaxies show a similar discrepancy at their respective outermost measurements. For DDO 154, the observed velocity is 45.50 km/s and the baryonic prediction is 16.25 km/s at 5.92 kpc, giving a discrepancy of 29.25 km/s. For NGC 2683, the corresponding velocities are 151.00 km/s and 73.94 km/s at 34.62 kpc, giving a discrepancy of 77.06 km/s.

These outermost measurements should not be used alone to rank the galaxies because their radial coverage, structures, and sampling differ. Instead, the rotation curves and discrepancy plots should be considered together to assess how the difference varies across the observed radii.

The results are consistent with the possibility of additional gravitating matter beyond the modelled baryonic contribution. However, this analysis does not independently determine a dark-matter density profile or rule out other explanations. The findings depend on the adopted mass-to-light ratios, observational uncertainties, and the limitations of the model.

## 8. Limitations

This analysis has several limitations.

1. **Mass-to-light ratio assumptions:** The predicted stellar contribution depends on the adopted mass-to-light ratios. Different values would change the calculated baryonic rotation curve.

2. **Observational uncertainties:** The plotted error bars represent uncertainties in the observed velocities but do not account for every source of uncertainty in the model.

3. **Galaxy geometry:** Real galaxies contain flattened disks and may have complex structures. The spherical circular-velocity relation introduced in the theoretical background is therefore illustrative rather than a complete description of the gravitational field of each galaxy.

4. **Limited sample size:** Only three galaxies are examined. The results cannot be generalised to all galaxies without a broader analysis of the SPARC sample.

5. **Scope of the model:** The analysis compares observed velocities with a predicted baryonic contribution. It does not fit a separate dark-matter halo model, determine a dark-matter density profile, or test alternative theories of gravity.

Consequently, the conclusions are restricted to whether the adopted visible-matter model reproduces the observed rotation curves of the selected galaxies and what the differences may imply.

This project investigates galactic rotation curves by comparing observed rotation velocities with the predicted contribution from gas and stellar components using data from the SPARC database. NGC 3198 is analysed as the main case study, while NGC 2683 and DDO 154 are used for comparison.

The baryonic rotation curves, velocity discrepancies, and observed-to-baryonic velocity ratios provide complementary ways to examine the agreement between the observed motion and the adopted visible-matter model. Where a persistent discrepancy is present, the model does not fully reproduce the observed rotation curve under the assumptions used.

Such discrepancies are consistent with the additional gravitational influence commonly attributed to dark matter. However, the calculations performed here do not independently establish the nature or distribution of the missing gravitational contribution. Overall, the project demonstrates how observational rotation-curve data can be used to investigate the relationship between visible matter and galactic dynamics.
